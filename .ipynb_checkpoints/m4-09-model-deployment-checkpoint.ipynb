```python
import numpy as np
import pandas as pd
import joblib
import requests
from sklearn.datasets import load_iris
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report
```

# Task 1: Train & Serialize

```python
iris = load_iris()
X_train, X_test, y_train, y_test = train_test_split(
    iris.data, iris.target, test_size=0.2, random_state=42
)

model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X_train, y_train)

accuracy = accuracy_score(y_test, model.predict(X_test))
print(f"Model accuracy: {accuracy:.4f}")
print(f"Feature names: {iris.feature_names}")
print(f"Target names: {list(iris.target_names)}")
print(f"classification_report: {classification_report(y_test, model.predict(X_test))}")
```

```python
joblib.dump(model, "model.joblib")
joblib.dump(iris.target_names, "target_names.joblib")
original_preds = model.predict(X_test)

loaded_model = joblib.load("model.joblib")

loaded_preds = loaded_model.predict(X_test)

accuracy = accuracy_score(y_test, loaded_preds)
print(f"Loaded model accuracy: {accuracy:.4f}")

are_equal = np.array_equal(original_preds, loaded_preds)
print(f"Predictions are identical: {are_equal}")
```

# Task 3: Test the API

```python
BASE_URL = "http://localhost:5000"
health_res = requests.get(f"{BASE_URL}/health")
print(f"1. Health Check: {health_res.status_code} - {health_res.json()}")

single_data = {"features": [5.1, 3.5, 1.4, 0.2]}
single_res = requests.post(f"{BASE_URL}/predict", json=single_data)
print(f"2. Single Prediction (200 OK): {single_res.json()}")

err_a = requests.post(f"{BASE_URL}/predict", json={})
print(f"Test A (Missing Key): Status {err_a.status_code}, Response: {err_a.json()}")

err_b = requests.post(f"{BASE_URL}/predict", json={"features": [5.1, 3.5, 1.4]})
print(f"Test B (Wrong Count): Status {err_b.status_code}, Response: {err_b.json()}")

err_c = requests.post(f"{BASE_URL}/predict", json={"features": ["a", "b", "c", "d"]})
print(f"Test C (Non-numeric): Status {err_c.status_code}, Response: {err_c.json()}")


batch_samples = X_test[:5].tolist()
batch_data = {"samples": batch_samples}

batch_res = requests.post(f"{BASE_URL}/predict_batch", json=batch_data)
print(f"4. Batch Status: {batch_res.status_code}")
target_list = list(iris.target_names)
api_preds = [target_list.index(p['predicted_class']) for p in batch_res.json()['predictions']]
local_preds = model.predict(X_test[:5])

match = np.array_equal(local_preds, api_preds)
print(f"5. Predictions match local model: {match}")
```

## Task 3: API Testing and Verification
In this section, we verify the Flask API's functionality using the `requests` library.
1. **Health Check**: Confirms the server is reachable.
2. **Single Prediction**: Tests the `/predict` endpoint with a single sample.
3. **Error Handling**: Validates that the API correctly rejects malformed inputs with a 400 error.
4. **Batch Prediction**: Ensures the API's batch results are identical to our local model's predictions.

# Task 4: Documentation & Reflection

1. Production Deployment
WSGI Server: Instead of the built-in Flask development server, use a production-grade WSGI server like Gunicorn or uWSGI for concurrency.

Containerization: Use Docker to package the application, ensuring consistency across different environments (local, staging, production).

Security: Implement HTTPS for encrypted communication and manage sensitive data (API keys, ports) using Environment Variables.

2. Model Versioning
Tracking: Maintain a versioned storage system (e.g., model_v1.joblib, model_v2.joblib) to allow quick rollbacks if a new model underperforms.

Deployment Strategy: Use A/B Testing or Canary Deployments to route a small percentage of traffic to the new model before a full rollout.

3. Monitoring
Latency: Track the time taken per prediction to ensure the API meets performance requirements.

Data Drift: Monitor input features for changes over time that might invalidate the model's training assumptions.

Error Rates: Log 4xx and 5xx HTTP errors to identify bugs or malformed client requests.

4. Handling High Traffic (Scalability)
Load Balancing: Use a load balancer (like Nginx or AWS ELB) to distribute incoming traffic across multiple server instances.

Horizontal Scaling: Deploy the API on a cluster (e.g., Kubernetes) to automatically scale the number of running containers based on demand.

Caching: Implement a caching layer (e.g., Redis) to store and reuse results for frequent, identical requests.
